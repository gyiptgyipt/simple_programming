# One of the most common error in deep learning: shape errors

```python
# Shapes for matrix multipliacation
tensor_A = torch.tensor([[1,2],
                        [3,4],
                        [5,6]])

tensor_B = torch.tensor([[7,10],
                        [8,11],
                        [9,12]])

torch.matmul(tensor_A, tensor_B.T)
```
