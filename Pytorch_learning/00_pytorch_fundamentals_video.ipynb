Tensor datatypes

```python
float_32_tenosr = torch.tensor([3.0, 6.0, 9.0],
                               dtype=None,
                               requires_grad=False)
float_32_tenosr
```

```python
float_16_tensor = float_32_tenosr.type(torch.float16)
float_16_tensor
```

```python
float_32_tenosr * float_16_tensor
```

```python
int_32_tensor = torch.tensor([3, 6, 9], dtype=torch.long)
int_32_tensor
```

```python
float_32_tenosr * int_32_tensor
```
